```python
# INNER JOIN
query = """
SELECT
    j.application_id,
    j.company_name,
    j.job_role,
    i.interview_round,
    i.interview_result
FROM job_applications AS j
INNER JOIN interviews AS i
ON j.application_id = i.application_id;
"""

result = pd.read_sql_query(query, conn)
display(result)
```

```python
# LEFT JOIN
# Find applications without interviews
query = """
SELECT
    j.application_id,
    j.company_name,
    j.job_role
FROM job_applications AS j
LEFT JOIN interviews AS i
ON j.application_id = i.application_id
WHERE i.application_id IS NULL"""
result=pd.read_sql_query(query,conn)
display(result)

```

```python
# RIGHT JOIN
query="""
SELECT
    j.company_name,
    i.application_id,
    i.interview_round
FROM job_applications AS j
RIGHT JOIN interviews AS i
ON j.application_id = i.application_id;"""
result=pd.read_sql_query(query,conn)
display(result)
```

```python
# FULL OUTER JOIN
query="""
SELECT
    j.application_id,
    j.company_name,
    i.interview_round,
    i.interview_result
FROM job_applications AS j
FULL OUTER JOIN interviews AS i
ON j.application_id = i.application_id;"""
result=pd.read_sql_query(query,conn)
display(result)

```

```python
# imagine your job applications have different salary ranges.
query = """
SELECT
    company_name,
    salary_lpa,
    CASE
        WHEN salary_lpa < 5 THEN 'Low'
        WHEN salary_lpa <= 6 THEN 'Medium'
        ELSE 'High'
    END AS Salary_Category
FROM job_applications;
"""

display(pd.read_sql_query(query, conn))
```

```python
# Can we classify job application progress?
query="""
SELECT
    company_name,
    shortlisted AS application_status,
    CASE
        WHEN shortlisted = 'Applied' THEN 'Waiting'
        WHEN shortlisted = 'Interview' THEN 'In Progress'
        WHEN shortlisted = 'Offer' THEN 'Successful'
        WHEN shortlisted = 'Rejected' THEN 'Unsuccessful'
        ELSE 'Other'
    END AS Application_Category
FROM job_applications;"""
display(pd.read_sql_query(query,conn))
```

```python
# How many applications are in each salary category?
query="""
SELECT
    CASE
        WHEN salary_lpa < 5 THEN 'Low'
        WHEN salary_lpa <= 6 THEN 'Medium'
        ELSE 'High'
    END AS Salary_Category,

    COUNT(*) AS Total_Applications

FROM job_applications

GROUP BY Salary_Category;"""
display(pd.read_sql_query(query,conn))
```

```python
# How many applications have reached the interview stage, and how many have been rejected?
query="""
SELECT
    COUNT(*) AS Total_Applications,

    SUM(CASE
        WHEN shortlisted = 'Interview' THEN 1
        ELSE 0
    END) AS Interview_Count,

    SUM(CASE
        WHEN shortlisted = 'Rejected' THEN 1
        ELSE 0
    END) AS Rejected_Count

FROM job_applications;"""
display(pd.read_sql_query(query,conn))
```
